### What is Pandas

Pandas is a fast, powerful, flexible and easy to use open source data analysis and manipulation tool,
built on top of the Python programming language.

https://pandas.pydata.org/about/index.html

### Pandas Series

A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

### Importing Pandas

```python
import numpy as np
import pandas as pd
```

### Series from lists

```python
# string
country = ['india', 'pakistan', 'usa', 'nepal', 'shrilanka']
pd.Series(country)
```

```python
# integers
runs = [31, 24, 78, 56, 100]

runs_series = pd.Series(runs)
```

```python
# custom index
marks = [67, 57, 89, 100]
subjects = ['maths', 'english', 'science', 'hindi']

pd.Series(marks, index=subjects)
```

```python
# setting a name
marks = pd.Series(marks, index=subjects, name="Lokendra's Marks")
print(marks)
```

### Series from dict

```python
marks = {
    'math': 67,
    'english': 57,
    'science': 89,
    'hindi': 100
}

marks_series = pd.Series(marks, name="Lokendra's Marks")
print(type(marks_series))
print(marks_series)
```
