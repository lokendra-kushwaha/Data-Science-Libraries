```python
# problem in vectorized opertions in vanilla python
s = ['cat','mat',None,'rat']

try:
    [i.startswith('c') for i in s]
except:
    print('An error occured!')
```

```python
# How pandas solves this issue?

s = pd.Series(['cat','mat',None,'rat'])

# string accessor
s.str.startswith('c')
#        ^
#        |
# fast and optimized
```

```python
# import titanic
df = pd.read_csv('../datasets/titanic.csv')
df['Name']
```

```python
# Common Functions ->

# lower/upper/capitalize/title
df['Name'].str.lower()
df['Name'].str.upper()
df['Name'].str.capitalize()
df['Name'].str.title()

# len
df[df['Name'].str.len() == 82].values[0]

# strip
'                   Lokendra                     '.strip()
df['Name'].str.strip()
```

```python
# split -> get
df['lastname'] = df['Name'].str.split(',').str.get(0)
df.head()

df[['title', 'firstname']] = df['Name'].str.split(',').str.get(1).str.strip().str.split(' ', n=1, expand=True)
df.head()


df['title'].value_counts()
```

```python
# replace

df['title'] = df['title'].str.replace('Ms.', 'Miss')
df['title'] = df['title'].str.replace('Mlle.', 'Miss')

df['title'].value_counts()
```

```python
# filtering

# startswith/endswith ->
df[df['firstname'].str.endswith('A')]
df[df['firstname'].str.endswith('z')]

# isdigit/isalpha ->
df[df['firstname'].str.isdigit()]
```

```python
# applying regex
# contains

# search john -> both case
df[df['firstname'].str.contains('john', case=False)] # case=False -> lower and upper both

# find lastnames with start and end char vowel
df[df['lastname'].str.contains('^[aeiouAEIOU].+[aeiouAEIOU]$')]

# find lastnames with start and end char consonants
df[df['lastname'].str.contains('^[^aeiouAEIOU].+[^aeiouAEIOU]$')]
```

```python
# slicing
df['Name'].str[:4]
df['Name'].str[::2]
df['Name'].str[::-1]
```
