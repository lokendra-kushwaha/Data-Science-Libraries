```python
import pandas as pd
import numpy as np

# ==============================================================================
# 🧪 EXPERIMENT: THE SILENT NaN MEMORY TRAP
# ==============================================================================

# Step 1: Create 1 Crore (10 Million) Integers
# Using 'int64' explicitly to standardize the test
N = 10_000_000
data_array = np.random.randint(1, 100, size=N, dtype=np.int64)

# Create a Pandas Series
s_clean = pd.Series(data_array)

print("--- BEFORE ADDING NaN (Pure Integers) ---")
print(f"Data Type : {s_clean.dtype}")
print(f"Memory Consumed : {s_clean.memory_usage(index=False) / (1024 * 1024):.2f} MB")

# ------------------------------------------------------------------------------
# Step 2: The Trap - Injecting just ONE missing value (NaN)
# ------------------------------------------------------------------------------
# We copy the series to avoid changing the original one
s_messy = s_clean.copy()

# Injecting NaN at the 5th position (index 4)
s_messy.iloc[4] = np.nan

print("\n--- AFTER ADDING JUST ONE NaN ---")
print(f"Data Type : {s_messy.dtype}")
print(f"Memory Consumed : {s_messy.memory_usage(index=False) / (1024 * 1024):.2f} MB")
```

```python
import pandas as pd
import numpy as np

# ==============================================================================
# 🚀 EXPERIMENT: THE MODERN PANDAS SOLUTION (Nullable Integers)
# ==============================================================================

N = 10_000_000
data_array = np.random.randint(1, 100, size=N)

# Notice the capital "I" in "Int64"
# This tells Pandas to use the modern Nullable Integer Extension Array
s_smart = pd.Series(data_array, dtype="Int64")

print("--- BEFORE ADDING NaN (Smart Integers) ---")
print(f"Data Type : {s_smart.dtype}")
print(f"Memory Consumed : {s_smart.memory_usage(index=False) / (1024 * 1024):.2f} MB")

# ------------------------------------------------------------------------------
# Injecting NaN to test if it survives without converting to float
# ------------------------------------------------------------------------------
s_smart_messy = s_smart.copy()
s_smart_messy.iloc[4] = np.nan

print("\n--- AFTER ADDING NaN (Did it survive?) ---")
print(f"Data Type : {s_smart_messy.dtype}")
print(f"Memory Consumed : {s_smart_messy.memory_usage(index=False) / (1024 * 1024):.2f} MB")
```
