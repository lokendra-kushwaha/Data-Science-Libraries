```python
import pandas as pd

# ---------------------------------------------------------
# SMART LOADING: Ignore useless columns to save memory
# ---------------------------------------------------------

# We explicitly tell Pandas which columns to extract from the CSV.
# It will completely ignore heavy string columns like 'email', 'phone', 'address'.
columns_we_need = ['user_id', 'age', 'subscription_plan', 'total_spent']

# The 'usecols' parameter prevents unwanted data from ever entering the RAM
try:
    df = pd.read_csv('massive_users_data.csv', usecols=columns_we_need)
except:
    print('Some error occured!')

# Now, 'df' only contains 4 columns instead of maybe 50!
print("Data loaded successfully!")
try:
    print(df.info())
except:
    print("Some error occured!")
```
