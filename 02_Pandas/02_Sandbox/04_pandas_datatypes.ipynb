```python
import pandas as pd
import numpy as np

# 1. We create a pure NumPy array (The Engine)
pure_numpy_array = np.array([10, 20, 30, 40, 50])

# 2. We put this array inside a Pandas Series (Building the Car)
my_series = pd.Series(pure_numpy_array)

# Check the outer shell (The Car)
print("Outer Object Type :", type(my_series))
# Output: <class 'pandas.core.series.Series'>

# 3. OPENING THE HOOD: Let's look inside the Series using .values
hidden_engine = my_series.values

# Check the inner engine
print("Inner Object Type :", type(hidden_engine))
# Output: <class 'numpy.ndarray'> 🤯
```
