```python
import pandas as pd
import numpy as np
import time

# ==============================================================================
# 🧪 EXPERIMENT 1: THE MEMORY MYTH (RangeIndex vs Data Values)
# ==============================================================================
print("--- EXPERIMENT 1: MEMORY USAGE ---")

# Generating 1 Crore (10 Million) random numbers
N = 10_000_000
data = np.random.randint(10000, 90000, size=N)

# Creating a Pandas Series (It automatically creates a RangeIndex in the background)
series_a = pd.Series(data)
series_b = pd.Series(np.random.randint(1000, 5000, size=N))

# Calculating actual memory consumed in RAM
index_memory_bytes = series_a.index.memory_usage() 
data_memory_mb = series_a.memory_usage(index=False) / (1024 * 1024)

print(f"Memory used by Index (RangeIndex for 10M rows) : {index_memory_bytes} Bytes only!")
print(f"Memory used by Data  (The actual 10M values)   : {data_memory_mb:.2f} MB")


# ==============================================================================
# 🚀 EXPERIMENT 2: PANDAS ALIGNMENT vs NUMPY SIMD
# ==============================================================================
print("\n--- EXPERIMENT 2: CALCULATION SPEED ---")

# TEST 1: The Pandas Way (Checks alignment, creates hash maps, then calculates)
start_time = time.perf_counter()
result_pandas = series_a + series_b
pandas_time = time.perf_counter() - start_time
print(f"Pandas Addition Time (With Alignment overhead) : {pandas_time:.5f} seconds")

# TEST 2: The NumPy Way (Bypasses Pandas index, directly uses CPU SIMD)
start_time = time.perf_counter()
result_numpy = series_a.to_numpy() + series_b.to_numpy()
numpy_time = time.perf_counter() - start_time
print(f"NumPy Addition Time  (Pure SIMD Execution)     : {numpy_time:.5f} seconds")

# -----------------------------------------
# 📊 FINAL VERDICT
# -----------------------------------------
speed_difference = pandas_time / numpy_time
print(f"\n💥 VERDICT: NumPy SIMD was {speed_difference:.2f}x FASTER than Pandas!")
```
