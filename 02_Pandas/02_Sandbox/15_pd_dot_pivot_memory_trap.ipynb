```python
import pandas as pd

# 1. Pure integer data (Fast & Memory Efficient)
df_raw = pd.DataFrame({
    'Day': ['Mon', 'Tue', 'Mon'],
    'Company': ['Apple', 'Apple', 'Google'],
    'Price': [150, 155, 2800] # Perfect int64
})

print("--- Before Pivot (Memory Types) ---")
print(df_raw.dtypes)

# 2. Pivoting introduces a NaN (Because Google didn't trade on Tue)
df_pivot = df_raw.pivot(index='Day', columns='Company', values='Price')

print("\n--- After Pivot (The NaN Penalty) ---")
print(df_pivot)

print("\n--- New Memory Types (Forced Upcast!) ---")
print(df_pivot.dtypes) # Watch the int64 become a slow float64
```
