```python
import pandas as pd
import numpy as np

# 1. Create a simple Datetime Series
df = pd.DataFrame({
    'Timestamp': pd.to_datetime(['1970-01-01 00:00:00', '1970-01-01 00:00:01', '2026-10-04 12:00:00'])
})

print("--- The User View (Illusion) ---")
print(df['Timestamp'])

print("\n--- The Hardware View (Raw RAM) ---")
# .to_numpy() strips the Pandas Series wrapper
# .view('int64') reveals the raw memory block in C
raw_memory = df['Timestamp'].to_numpy().view('int64')
print(raw_memory)

# """
# Output of Raw Memory:
# 0                             0  <- 0 nanoseconds from Epoch
# 1                    1000000000  <- 1 billion nanoseconds = 1 second
# 2           1791115200000000000  <- Trillions of nanoseconds for 2026!
# """
```
