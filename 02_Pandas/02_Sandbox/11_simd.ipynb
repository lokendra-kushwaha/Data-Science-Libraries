```python
import pandas as pd
import numpy as np
import time

# ==============================================================================
# 🏎️ EXPERIMENT: THE GIL TRAP vs SIMD VECTORIZATION
# ==============================================================================

N = 10_000_000  # 1 Crore numbers

# Creating two arrays of pure C-integers
array1 = np.random.randint(1, 100, size=N, dtype=np.int32)
array2 = np.random.randint(1, 100, size=N, dtype=np.int32)

# ==============================================================================
# 1. THE PYTHON LOOP (The Sin of Big Data)
# ==============================================================================
print("1. Running Python 'for' loop (Please wait, this will be very slow)...")

# Converting to Python lists to simulate pure Python behavior
list1 = array1.tolist()
list2 = array2.tolist()
result_list = []

start_time = time.time()
for i in range(N):
    result_list.append(list1[i] + list2[i])
loop_time = time.time() - start_time

print(f"Time taken by Python Loop: {loop_time:.4f} seconds")

# ==============================================================================
# 2. PANDAS VECTORIZATION (C-Backend)
# ==============================================================================
print("\n2. Running Pandas Vectorization...")
s1 = pd.Series(array1)
s2 = pd.Series(array2)

start_time = time.time()
result_series = s1 + s2  # Direct addition (No loops visible)
pandas_time = time.time() - start_time

print(f"Time taken by Pandas: {pandas_time:.4f} seconds")

# ==============================================================================
# 3. PURE NUMPY VECTORIZATION (Direct SIMD Hardware Access)
# ==============================================================================
print("\n3. Running Pure NumPy Vectorization...")

start_time = time.time()
result_array = array1 + array2  # Bypassing Python entirely
numpy_time = time.time() - start_time

print(f"Time taken by NumPy: {numpy_time:.4f} seconds")

# ==============================================================================
# CONCLUSION
# ==============================================================================
print("-" * 50)
print(f"🚀 NumPy is approximately {loop_time / numpy_time:.0f}x FASTER than the Python Loop!")
```
