```python
import pandas as pd
import numpy as np

# Simulating a large dataset with repetitive hierarchical data
countries = ['India'] * 10000 + ['USA'] * 10000
years = [2023, 2024] * 10000

# Creating the MultiIndex
mi = pd.MultiIndex.from_arrays([countries, years], names=['Country', 'Year'])
df = pd.DataFrame({'Sales': np.random.randint(100, 1000, 20000)}, index=mi)

print("--- The Hardware Reality of MultiIndex ---")
print("1. Unique Strings Stored in RAM (levels):")
print(df.index.levels[0])

print("\n2. What the CPU Actually Reads (codes/pointers):")
print(df.index.codes[0][:15], "... (array continues)")
print("DataType of Pointers:", df.index.codes[0].dtype) # Notice it's an efficient int8
```
