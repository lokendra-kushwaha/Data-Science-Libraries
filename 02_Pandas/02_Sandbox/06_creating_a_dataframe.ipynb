```python
import pandas as pd

# ---------------------------------------------------------
# CREATING OUR FIRST DATAFRAME (2D Table)
# ---------------------------------------------------------

# Using a Dictionary to create a DataFrame. 
# Keys become Column Names, and Lists become the data.
data = {
    'Name': ['Amit', 'Priya', 'Rahul', 'Neha'],         # String data
    'Age': [25, 28, 22, 30],                            # Integer data
    'Salary': [55000.50, 72000.00, 48000.75, 90000.00]  # Float data
}

df = pd.DataFrame(data)

# Displaying the beautiful 2D table
print("--- Our DataFrame ---")
print(df)

# The Magic: Let's check the data types of each column
print("\n--- Data Types of each Column ---")
print(df.dtypes)
```

```python
import pandas as pd

# 1. The Wall (DataFrame)
data = {
    'Age': [25, 30, 22], 
    'Salary': [50000, 60000, 45000]
}
df = pd.DataFrame(data)

# 2. Extracting one column gives you a Series (The Brick)
age_column = df['Age']
print(f"Type of extracted column: {type(age_column)}\n")
# Output: <class 'pandas.core.series.Series'>

# 3. Performing Math on DataFrame returns a Series
max_values = df.max()
print("--- Maximum values of each column ---")
print(max_values)
print(f"\nType of result: {type(max_values)}")
# Output: <class 'pandas.core.series.Series'>
```
