```python
import pandas as pd

# =====================================================================
# ❌ THE WRONG WAY (What I imagined)
# Passing a dictionary directly to a Series
# =====================================================================
wrong_series = pd.Series({'Age': [25, 30, 22]})

print("--- Wrong Series ---")
print(wrong_series)
print(f"Dtype: {wrong_series.dtype}") 
# Output: object (It crammed the whole list into ONE row. SIMD is dead!)


# =====================================================================
# ✅ THE RIGHT WAY (How DataFrame actually builds it in the background)
# Data is passed as a list, and 'Age' is passed as the 'name' attribute
# =====================================================================
correct_series = pd.Series([25, 30, 22], name='Age')

print("\n--- Correct Series ---")
print(correct_series)
print(f"Dtype: {correct_series.dtype}") 
# Output: int64 (Data is spread vertically. Pure C-Array, SIMD ready!)
```
