```python
import pandas as pd

# Creating a wide format dataframe
df_wide = pd.DataFrame({
    '2023': [500, 700],
    '2024': [600, 750]
}, index=['India', 'USA'])

print("--- unstack() / Wide Format (Human Readable) ---")
print(df_wide)
print("Shape:", df_wide.shape)

# Stacking compresses columns into the index
df_long = df_wide.stack()

print("\n--- stack() / Long Format (Hardware Friendly 1D Block) ---")
print(df_long)
print("Shape:", df_long.shape)
```
