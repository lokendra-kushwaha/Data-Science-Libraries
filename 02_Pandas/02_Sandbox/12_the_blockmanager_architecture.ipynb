```python
import pandas as pd
import numpy as np

# ==============================================================================
# 🧱 EXPERIMENT: THE HIDDEN BLOCKMANAGER & FRAGMENTATION
# ==============================================================================

# 1. Create a clean DataFrame
# 3 int columns and 2 float columns
df = pd.DataFrame({
    'A': np.random.randint(1, 100, 10),
    'B': np.random.randint(1, 100, 10),
    'C': np.random.randint(1, 100, 10),
    'D': np.random.rand(10),
    'E': np.random.rand(10)
})

print("--- 1. Initial Architecture (Perfectly Fused Blocks) ---")
# Using the hidden private attribute '_mgr' to see the backend blocks
print(df._mgr)

# ------------------------------------------------------------------------------
# 2. Shattering the Architecture (Memory Fragmentation)
# Adding new columns randomly causes Pandas to create isolated blocks
df['F'] = np.random.randint(1, 100, 10)  
df['G'] = np.random.rand(10)             
df['H'] = np.random.randint(1, 100, 10)  

print("\n--- 2. Fragmented Architecture (Scattered RAM) ---")
print(df._mgr)

# ------------------------------------------------------------------------------
# 3. Forcing Consolidation
# We manually tell Pandas to defragment and fuse the blocks back together
df = df.copy() # In modern Pandas, copying triggers consolidation 

print("\n--- 3. After Consolidation (Fused & Optimized again) ---")
print(df._mgr)
```

```python
import pandas as pd
import numpy as np

# ==============================================================================
# 🚀 THE BIG DATA WAY: APPENDING 1 MILLION ROWS TO A 10GB FILE
# ==============================================================================

# WARNING: We have a '10gb_massive_data.csv' file on our Hard Drive.
# DO NOT run pd.read_csv('10gb_massive_data.csv'). It will crash the RAM immediately!

# ------------------------------------------------------------------------------
# STEP 1: Generate/Prepare the new 1 Million Rows
# This is a small dataset (~15 MB), so it easily fits in our 8GB RAM.
# ------------------------------------------------------------------------------
print("Preparing new 1 Million rows in RAM...")
new_1m_rows = pd.DataFrame({
    'A': np.random.randint(1, 100, 1_000_000),
    'B': np.random.randint(1, 100, 1_000_000)
})

# ------------------------------------------------------------------------------
# STEP 2: The Hardware Hack (Disk-Level Appending)
# ------------------------------------------------------------------------------
# We do not touch the old 10GB data. We just tell Pandas to open the Hard Drive file
# in 'append' mode (mode='a') and write these new rows at the very bottom.
# header=False ensures we don't write column names again in the middle of the file.

print("Writing directly to the Hard Drive...")
new_1m_rows.to_csv('10gb_massive_data.csv', mode='a', header=False, index=False)

print("✅ Success! 1 Million rows seamlessly added to the 10GB file without crashing RAM.")
```
