```python
import pandas as pd
import numpy as np

# ==============================================================================
# OUT-OF-CORE PROCESSING: HANDLING MASSIVE DATASETS (e.g., 50GB CSV)
# ==============================================================================

# Simulating a massive dataset scenario
# Instead of loading the entire file, we define a 'chunksize'.
# This tells Pandas to only load 100,000 rows into RAM at a single time.

file_path = "../01_Core_Modules/datasets/movies.csv"
chunk_size = 100_000 

# 'pd.read_csv' now returns a Generator object (TextFileReader), not a DataFrame
chunk_generator = pd.read_csv(file_path, chunksize=chunk_size)

print("Started streaming data from the hard drive...\n")

total_rows_processed = 0
total_sales = 0.0

# Iterating through the generator. 
# Memory is cleared and reused for each iteration (Lazy Evaluation).
for chunk_number, chunk_df in enumerate(chunk_generator, start=1):
    
    # Process only the current chunk residing in RAM
    current_chunk_rows = len(chunk_df)
    total_rows_processed += current_chunk_rows
    
    # Simulating a mathematical operation (e.g., calculating total sales)
    if 'Sales' in chunk_df.columns:
        total_sales += chunk_df['Sales'].sum()
    
    print(f"Chunk {chunk_number} Processed | RAM loaded: {current_chunk_rows} rows | Total Processed: {total_rows_processed}")

print("\nData processing complete. RAM remained completely safe!")
```
