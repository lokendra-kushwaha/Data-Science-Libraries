```python
import pandas as pd

# Wide Data (Bad for Machine Learning / SIMD)
df_wide = pd.DataFrame({
    'Store_ID': ['S-101', 'S-102'],
    'Jan_Sales': [5000, 7000],
    'Feb_Sales': [5200, 7100],
    'Mar_Sales': [4900, 7300]
})

print("--- Original Wide Data (Scattered Features) ---")
print(df_wide)

# Melted Data (SIMD & ML Optimized)
df_melted = pd.melt(
    df_wide,
    id_vars=['Store_ID'],
    var_name='Month',
    value_name='Target_Sales' # Creating the Y variable
)

print("\n--- Melted Data (Single Contiguous Target for ML) ---")
print(df_melted)
```
