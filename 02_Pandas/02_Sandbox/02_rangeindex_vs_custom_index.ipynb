```python
import pandas as pd
import numpy as np

N = 10_000_000
data = np.random.randint(1, 100, size=N)

# ------------------------------------------------
# CASE 1: Trusting Pandas (Default Index)
# ------------------------------------------------
s_trust = pd.Series(data)
print(f"Index Type: {type(s_trust.index)}")  
# Output: <class 'pandas.core.indexes.range.RangeIndex'>
print(f"Memory: {s_trust.index.memory_usage()} Bytes")  
# Output: 132 Bytes 🤩

# ------------------------------------------------
# CASE 2: Doubting Pandas (Independent Custom Index)
# ------------------------------------------------
my_index = np.arange(N) # We explicitly created this 0 to 10M array
s_doubt = pd.Series(data, index=my_index)

print(f"\nIndex Type: {type(s_doubt.index)}")  
# Output: <class 'pandas.core.indexes.numeric.Int64Index'> 
print(f"Memory: {s_doubt.index.memory_usage() / (1024 * 1024):.2f} MB")  
# Output: ~76.29 MB 😱
```
