```python
import time

all_dfs = []

for j in range(1, 501):
    url = f'https://www.ambitionbox.com/list-of-companies?page={j}'
    headers = {'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
    webpage = requests.get(url, headers=headers).text

    time.sleep(2)

    soup = BeautifulSoup(webpage, 'lxml')
    company = soup.find_all('div', class_="companyCardWrapper")

    name = []
    rating = []
    review = []
    c_type = []
    hq = []
    salaries = []
    interviews = []
    jobs = []
    benefits = []

    for info in company:
        # Name
        name_tag = info.find('h2')
        name.append(name_tag.text.strip() if name_tag else "N/A")
        
        # Rating
        rating_tag = info.find('div', style="height:auto;")
        rating.append(rating_tag.text.strip() if rating_tag else "N/A")
        
        # Review
        review_tag = info.find('span', class_="companyCardWrapper__companyRatingCount")
        review.append(review_tag.text.strip() if review_tag else "N/A")
        
        # Company Type and Headquarter (split handling)
        interlink_tag = info.find('span', class_="companyCardWrapper__interLinking")
        if interlink_tag:
            parts = interlink_tag.text.strip().split(' | ')
            c_type.append(parts[0].strip() if len(parts) > 0 else "N/A")
            hq.append(parts[1].split(' +')[0].strip() if len(parts) > 1 else "N/A")
        else:
            c_type.append("N/A")
            hq.append("N/A")
            
        # Salaries, Interviews, Jobs, Benefits (Index mapping check)
        actions = info.find_all('span', class_="companyCardWrapper__ActionCount")
        salaries.append(actions[1].text.strip() if len(actions) > 1 else "N/A")
        interviews.append(actions[2].text.strip() if len(actions) > 2 else "N/A")
        jobs.append(actions[3].text.strip() if len(actions) > 3 else "N/A")
        benefits.append(actions[4].text.strip() if len(actions) > 4 else "N/A")
            
    d = {'Name': name, 'Rating': rating, 'Reviews': review, 'Type': c_type, 'Headquater': hq, 'Salaries': salaries, 'Interviews': interviews, 'Jobs': jobs, 'Benefits': benefits}
    df = pd.DataFrame(d)
    
    all_dfs.append(df)

final = pd.concat(all_dfs, ignore_index=True)
```

```python
final.to_csv('company_data.csv', index=False)
```
