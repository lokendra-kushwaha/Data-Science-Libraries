```python
import numpy as np

# Define the Action Matrix (Space Transformer)
A = np.array([[3, 1],
              [0, 2]])

# Extract Eigenvalues (Stretch factor) and Eigenvectors (Stubborn arrows)
eigenvalues, eigenvectors = np.linalg.eig(A)

print("=== THE STUBBORN ARROWS (EIGENVECTORS) ===")
print(eigenvectors)

print("\n=== THE STRETCH FACTOR (EIGENVALUES) ===")
print(eigenvalues)

# --- THE ULTIMATE PROOF TEST ---
# Equation to prove: A * v = Lambda * v

# Extract the first eigenvector (First column)
v1 = eigenvectors[:, 0]

# 1. Apply the Matrix Action (A dot v1)
action_result = np.dot(A, v1)

# 2. Apply the Scalar Stretch (Lambda * v1)
scalar_result = eigenvalues[0] * v1

print("\n=== PROOF: ACTION VS STRETCH ===")
print("Matrix Action (A x v1):      ", action_result)
print("Scalar Stretch (Lambda x v1):", scalar_result)
```
