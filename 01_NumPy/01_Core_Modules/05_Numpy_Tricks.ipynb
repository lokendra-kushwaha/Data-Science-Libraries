```python

# 25. np.allclose and equals

# Two arrays that are mathematically the same but have a tiny floating-point difference
arr1 = np.array([0.1 + 0.2])
arr2 = np.array([0.3])

# Check if they are exactly equal
exact_match = np.array_equal(arr1, arr2)
print("Exact match:", exact_match)

# Check if they are close enough (ignoring tiny decimal differences)
close_match = np.allclose(arr1, arr2)
print("Close match:", close_match)
```
