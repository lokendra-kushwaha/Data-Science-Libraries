### **What is numpy?**

**NumPy** is the fundamental package for scientific computing in Python. It is a Python library that provides a **multidimensional array object**, various derived objects (such as masked arrays and matrices), and an assortment of routines for fast operations on arrays, including:
* Mathematical and logical operations
* Shape manipulation
* Sorting and selecting
* I/O (Input/Output)
* Discrete Fourier transforms
* Basic linear algebra and basic statistical operations
* Random simulation and much more.

At the core of the NumPy package, is the **`ndarray`** object. This encapsulates n-dimensional arrays of homogeneous data types.

---

### **Numpy Arrays Vs Python Sequences**

* **Fixed Size:** NumPy arrays have a fixed size at creation, unlike Python lists (which can grow dynamically). Changing the size of an `ndarray` will create a new array and delete the original.
* **Homogeneous Elements:** The elements in a NumPy array are all required to be of the same data type, and thus will be the same size in memory.
* **Efficient Processing:** NumPy arrays facilitate advanced mathematical and other types of operations on large numbers of data. Typically, such operations are executed more efficiently and with less code than is possible using Python’s built-in sequences.
* **Scientific Ecosystem Standard:** A growing plethora of scientific and mathematical Python-based packages are using NumPy arrays; though these typically support Python-sequence input, they convert such input to NumPy arrays prior to processing, and they often output NumPy arrays.

### **1. Creating Numpy Arrays**
Below are the fundamental functions used to generate arrays in various dimensions and data types.

```python
import numpy as np

# np.array: Creates an array from a regular Python list or tuple
a = np.array([1, 2, 3]) # it is a 1D array
print("1D Array:\n", a)
print(type(a))
```

```python
# 2D and 3D array creation
b = np.array([[1, 2, 3], [4, 5, 6]])
print("\n2D Array:\n", b)

c = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print("\n3D Array:\n", c)
```

```python
# dtype: Forcing a specific data type during array creation
print("\nFloat array:", np.array([1, 2, 3], dtype=float))
print("Boolean array:", np.array([1, 2, 3], dtype=bool))
print("Complex array:", np.array([1, 2, 3], dtype=complex))
```

```python
# np.arange: Returns evenly spaced values within a given interval (start, stop, step)
print("\nArange (1 to 10):", np.arange(1, 11))
print("Arange with step of 2:", np.arange(1, 11, 2))
```

```python
# reshape: Gives a new shape to an array without changing its data
print("\nReshape to 5x2:\n", np.arange(1, 11).reshape(5, 2))
print("Reshape to 4x3:\n", np.arange(1, 13).reshape(4, 3))
```

```python
# np.ones and np.zeros: Initialize arrays filled entirely with 1s or 0s
print("\nZeros (3x4):\n", np.zeros((3, 4)))
print("Ones (3x4):\n", np.ones((3, 4)))
```

```python
# np.random: Generates an array with random floats between 0.0 and 1.0
print("\nRandom (3x4):\n", np.random.random((3, 4)))
```

```python
# np.linspace: Returns evenly spaced numbers over a specified interval (start, stop, number_of_samples)
print("\nLinspace (10 values from -10 to 10):\n", np.linspace(-10, 10, 10))
```

```python

# np.identity: Returns a square identity matrix (1s on the main diagonal, 0s elsewhere)
print("\nIdentity Matrix (3x3):\n", np.identity(3))
```

### **2. Array Attributes**
Attributes allow you to check the metadata of an array (its shape, size, memory consumption) without changing its underlying data.

```python
a1 = np.arange(10, dtype=np.int32)
a2 = np.arange(12, dtype=float).reshape(3, 4)
a3 = np.arange(8).reshape(2, 2, 2)
```

```python
# ndim: Returns the number of dimensions (axes) of the array
print("Dimensions of a1, a2, a3:", a1.ndim, a2.ndim, a3.ndim)
```

```python
# shape: Returns a tuple representing the size of the array in each dimension
print("Shapes of a1, a2, a3:", a1.shape, a2.shape, a3.shape)
```

```python
# size: Returns the total number of elements in the array
print("Total size of a1, a2, a3:", a1.size, a2.size, a3.size)
```

```python
# itemsize: Returns the length of one array element in bytes
print("Itemsize (bytes per element):", a1.itemsize, a2.itemsize, a3.itemsize)
```

```python
# dtype: Returns the data-type of the array's elements
print("Datatypes:")
print("a1:", a1.dtype)
print("a2:", a2.dtype)
print("a3:", a3.dtype)
```

```python
# ndarray.nbytes: Returns the total bytes consumed by the elements of the array
print("Total memory footprint (bytes):")
print("a1 nbytes:", a1.nbytes)
print("a2 nbytes:", a2.nbytes)
print("a3 nbytes:", a3.nbytes)
```

### **3. Changing Datatype**
You can cast arrays to new datatypes.

```python
# astype: Casts a copy of the array to a specified type
print("Original a3 dtype:", a3.dtype)
a3 = a3.astype(np.int32)
print("Changed a3 dtype:", a3.dtype)
```

### **4. Array Operations**
NumPy also allows mathematical operations directly on arrays (vectorization) without writing `for` loops.

```python
print("\n--- Array Operations ---")
a1 = np.arange(12).reshape(3, 4)
a2 = np.arange(12, 24).reshape(3, 4)
```

```python
# Scalor Operations (Operations between an array and a single number)
print("\nScalar Arithmetic (a1 ** 2):\n", a1 ** 2)
```

```python
# Relational Operations (Returns a boolean array)
print("\nRelational (a2 == 15):\n", a2 == 15)
```

```python
# Vector Operations (Operations between two arrays of the same shape)
print("\nVector Arithmetic (Addition):\n", a1 + a2)
print("\nVector Arithmetic (Subtraction):\n", a1 - a2)
print("\nVector Arithmetic (Exponentiation):\n", a1 ** a2)
```

### **5. Array Functions**
NumPy provides highly optimized built-in math functions. 
**Note on the `axis` parameter:**
* `axis=0` performs the operation down the **columns**.
* `axis=1` performs the operation across the **rows**.

```python
a1 = np.random.random((3, 3))
a1 = np.round(a1*100)
print("Base Array (a1):\n", a1)
```

```python
# max / min / sum / prod (Global operations on the whole array)
print("\nGlobal Max:", np.max(a1))
print("Global Min:", np.min(a1))
print("Global Sum:", np.sum(a1))
print("Global Product:", np.prod(a1))

# 0 -> col and 1 -> row (Axis-wise operations)
print("\nMax in each row (axis=1):", np.max(a1, axis=1))
print("Max in each col (axis=0):", np.max(a1, axis=0))
print("Min in each col (axis=0):", np.min(a1, axis=0))
print("Sum of each row (axis=1):", np.sum(a1, axis=1))
print("Product of each row (axis=1):", np.prod(a1, axis=1))
```

```python
# mean / median / std / var (Statistical operations)
print("\nOverall Mean:", np.mean(a1))
print("Mean of rows (axis=1):", np.mean(a1, axis=1)) 
print("Overall Median:", np.median(a1))
print("Standard Deviation:", np.std(a1)) 
print("Variance:", np.var(a1)) 
```

```python
# Trigonometric functions (Applies function to every element)
print("\nSine values:\n", np.sin(a1))
# print(np.cos(a1)) # Works identically
# print(np.tan(a1)) # Works identically
```

```python
# dot product (Matrix Multiplication)
a2 = np.arange(12).reshape(3, 4)
a3 = np.arange(12, 24).reshape(4, 3)
print("\nDot Product of a2 (3x4) and a3 (4x3):\n", np.dot(a2, a3))
```

```python
# log and exponents
print("\nNatural Logarithm:\n", np.log(a1))
print("Exponential (e^x):\n", np.exp(a1))
```

```python
# round / floor / ceil
random_arr = np.random.random((2, 3)) * 100
print("\nOriginal Random Array:\n", random_arr)
print("Rounded (nearest integer):\n", np.round(random_arr))
print("Floor (rounded down):\n", np.floor(random_arr))
print("Ceil (rounded up):\n", np.ceil(random_arr))
```
