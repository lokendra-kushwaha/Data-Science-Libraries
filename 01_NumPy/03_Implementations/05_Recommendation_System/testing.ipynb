```python
from data_cleaner import prepare_movie_data
from vectorizer import create_vector_matrix
from recommender import get_recommendations

print("🚀 Initializing Recommendation Engine ...")

cleaned_data = prepare_movie_data('movies.csv')
vector_matrix = create_vector_matrix(cleaned_data)

avatar_recs = get_recommendations('Avatar', cleaned_df=cleaned_data, vector_matrix=vector_matrix, top_n=3)
if avatar_recs:
    print("\n🍿 If you liked Avatar, We Recommend:")
    for movie in avatar_recs:
        print(f"  🎬 {movie}")

titanic_recs = get_recommendations('kaju ktli', cleaned_df=cleaned_data, vector_matrix=vector_matrix, top_n=3)
if titanic_recs:
    print("\n🍿 If you liked Titanic, We Recommend:")
    for movie in titanic_recs:
        print(f"  🎬 {movie}")
```
