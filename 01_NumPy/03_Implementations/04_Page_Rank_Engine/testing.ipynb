```python
# Cell 1: Import modules and load data into RAM
from crawler import build_web_graph
from page_rank import generate_transition_matrix, calculate_pagerank
from search_engine import search
import time

my_seeds = [
    "https://en.wikipedia.org/wiki/Data_science",
    "https://en.wikipedia.org/wiki/Artificial_intelligence"
]

start = time.time()
graph, index = build_web_graph(seed_urls=my_seeds, max_pages=70)
print(time.time() - start)
matrix, pages = generate_transition_matrix(graph)
ranks = calculate_pagerank(matrix, len(pages))
print("Engine Loaded in Memory!")
```

```python
# Cell 2: Search Query
results = search("is", index, pages, ranks)
for i, (url, score) in enumerate(results[:5]):
    print(f"{i+1}. {url} (Score: {score:.4f})")
```

```python
# Cell 3: Search Query
results = search("poetry", index, pages, ranks)
for i, (url, score) in enumerate(results[:5]):
    print(f"{i+1}. {url} (Score: {score:.4f})")
```
