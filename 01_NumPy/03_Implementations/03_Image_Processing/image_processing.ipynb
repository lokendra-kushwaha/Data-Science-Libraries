```python
import numpy as np
import matplotlib.pyplot as plt

image_matrix = plt.imread(r"C:\Users\loken\OneDrive\Pictures\01.jpg")

print("Shape:", image_matrix.shape)
print("Data type:", type(image_matrix))


print(image_matrix[0][0]) 

plt.imshow(image_matrix)
plt.axis('off')
plt.show()
```

```python
reverse_image = np.transpose(image_matrix, (1, 0, 2))

print("Shape:", reverse_image.shape)
print("Data type:", type(reverse_image))

print(reverse_image[0][0]) 

plt.imshow(reverse_image)
plt.axis('off')
plt.show()
```

```python
cropped_image = image_matrix[500:, 1000:]

print("Shape:", cropped_image.shape)
print("Data type:", type(cropped_image))

print(cropped_image[0][0]) 

plt.imshow(cropped_image)
plt.axis('off')
plt.show()
```

```python
normalized_image = image_matrix / 255.0

print("Shape:", normalized_image.shape)
print("Data type:", type(normalized_image))

print(normalized_image[0][0]) 

plt.imshow(normalized_image)
plt.axis('off')
plt.show()
```

```python
flattened_matrx = image_matrix.flatten()

print("Shape:", flattened_matrx.shape)
print("Data type:", type(flattened_matrx))

print(flattened_matrx) 
```

```python
red_channel = image_matrix[:, :, 0]

print("Shape:", red_channel.shape)
print("Data type:", type(red_channel))

print(red_channel) 
```

```python
flipped_image = np.flipud(image_matrix) # np.fliplr(image_matrix)

print("Shape:", flipped_image.shape)
print("Data type:", type(flipped_image))

print(flipped_image[0][0]) 

plt.imshow(flipped_image)
plt.axis('off')
plt.show()
```
